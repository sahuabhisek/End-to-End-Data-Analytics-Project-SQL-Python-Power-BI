```python
plt.figure(figsize=(12, 6))

# 3. Seaborn countplot ka use karke Grouped Bar Chart banana
sns.barplot(
    data=df,
    y="Payment Method",  # Y-axis par items rakhenge taaki naam saaf padhe ja sakein
    x="Gender",  # Male aur Female ke hisaab se bars alag-alag honge
    palette="Set2",
)

# 4. Chart styling aur labels
plt.title(
    "Comparison of Items Purchased by Gender",
    fontsize=16,
    fontweight="bold",
    pad=15,
)


# Gridlines enable karein better readability ke liye
plt.grid(axis="x", linestyle="--", alpha=0.7)

# Layout adjust karein taaki text cut na ho
plt.tight_layout()

# Chart ko high-resolution image ke roop mein save karein
plt.savefig("gender_vs_item_purchased.png", dpi=300)
print("Chart successfully saved as 'gender_vs_item_purchased.png'!")

# Display chart
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns



# Column names ke extra spaces/BOM ko clean karna
df.columns = df.columns.str.replace("ï»¿", "").str.strip()

# 2. Figure size set karein (Kyunki items zyada hain, toh width badi rakhein)
plt.figure(figsize=(14, 8))

# 3. Seaborn countplot ka use karke Grouped Bar Chart banana
sns.barplot(
    data=df,
    y="Category",  # Y-axis par items rakhenge taaki naam saaf padhe ja sakein
    hue="Gender",  # Male aur Female ke hisaab se bars alag-alag honge
    palette="Set2",
)

# 4. Chart styling aur labels
plt.title(
    "Comparison of Items Purchased by Gender",
    fontsize=16,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Count of Purchases", fontsize=12)
plt.ylabel("Item Purchased", fontsize=12)
plt.legend(title="Gender", loc="upper right")

# Gridlines enable karein better readability ke liye
plt.grid(axis="x", linestyle="--", alpha=0.7)

# Layout adjust karein taaki text cut na ho
plt.tight_layout()

# Chart ko high-resolution image ke roop mein save karein
plt.savefig("gender_vs_item_purchased.png", dpi=300)
print("Chart successfully saved as 'gender_vs_item_purchased.png'!")

# Display chart
plt.show()
```
